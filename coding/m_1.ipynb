```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_classification

# 1. Sun’iy ma’lumotlar yaratish
X, y = make_classification(
    n_samples=1000,
    n_features=5,
    n_informative=4,
    n_redundant=1,
    random_state=42
)

# DataFrame yaratamiz
df = pd.DataFrame(X, columns=["age", "income", "loan_amount", "credit_score", "employment_years"])
df["default"] = y

# 2. Ba’zi joylarda bo‘sh qiymatlar (NaN) kiritamiz
np.random.seed(42)
for col in ["age", "income", "loan_amount", "credit_score", "employment_years"]:
    df.loc[df.sample(frac=0.05).index, col] = np.nan  # 5% bo‘sh qiymatlar

# 3. Ba’zi joylarda ortiqcha (outlier) qiymatlar kiritamiz
outlier_indices = np.random.choice(df.index, 10, replace=False)
df.loc[outlier_indices, "income"] *= 10  # juda katta daromadlar

# 4. CSV faylga saqlash
df.to_csv("coding_challenge_clean_dataset.csv", index=False)

print("✅ Dataset yaratildi: coding_challenge_clean_dataset.csv")
print(df.head())
```
